Note: This dataset is simulated, not collected from a real job portal. It is generated with Python and NumPy using random values and a fixed seed (42).

```python
import pandas as pd
import numpy as np

# Set seed for consistent data generation
np.random.seed(42)

# Configuration Setup
total_records = 1500
years = [2021, 2022, 2023, 2024, 2025, 2026]
roles = [
    'Web Developer',
    'Data Analyst',
    'Software Engineer',
    'Data Entry Operator',
    'Manual QA Tester',
    'AI Prompt Engineer'
]

dataset = []

for _ in range(total_records):
    year = np.random.choice(years)
    role = np.random.choice(roles)

    # Handle the newly emerged role post-AI era
    if year <= 2022 and role == 'AI Prompt Engineer':
        role = 'Software Engineer'  # Replace because this role didn't exist back then

    # Logic Phase 1: Before AI Era (2021 - 2022)
    if year <= 2022:
        degree_importance = np.random.choice(['High', 'Medium'], p=[0.7, 0.3])
        portfolio_weightage = np.random.randint(30, 50)

        if role in ['Data Entry Operator', 'Manual QA Tester']:
            vacancies = np.random.randint(120, 180)
            hiring_difficulty = np.random.choice(['Easy', 'Medium'], p=[0.7, 0.3])
            top_skill = 'Excel & Typing' if role == 'Data Entry Operator' else 'Manual Test Cases'
            skill_demand_status = 'Stable'
            avg_salary_lpa = round(np.random.uniform(2.5, 4.0), 2)
        else:
            vacancies = np.random.randint(150, 250)
            hiring_difficulty = np.random.choice(['Easy', 'Medium', 'Hard'], p=[0.4, 0.5, 0.1])
            top_skill = 'HTML/CSS/JavaScript' if role == 'Web Developer' else ('SQL & Excel' if role == 'Data Analyst' else 'Java / C++')
            skill_demand_status = 'Growing'
            avg_salary_lpa = round(np.random.uniform(4.5, 7.5), 2)

    # Logic Phase 2: After AI Era (2023 - 2026)
    else:
        if role == 'AI Prompt Engineer':
            vacancies = np.random.randint(100, 200) if year >= 2024 else np.random.randint(10, 40)
            hiring_difficulty = np.random.choice(['Medium', 'Hard'], p=[0.4, 0.6])
            top_skill = 'LLMs & Prompt Engineering'
            skill_demand_status = 'High Growth'
            degree_importance = np.random.choice(['Medium', 'Low'], p=[0.3, 0.7])
            portfolio_weightage = np.random.randint(75, 95)
            avg_salary_lpa = round(np.random.uniform(8.0, 15.0), 2)

        elif role in ['Data Entry Operator', 'Manual QA Tester']:
            vacancies = np.random.randint(10, 45)  # Heavy drop due to automation
            hiring_difficulty = 'Hard'  # Very few openings left
            top_skill = 'Basic Data Entry' if role == 'Data Entry Operator' else 'Basic Manual Testing'
            skill_demand_status = 'Declining'
            degree_importance = 'Low'
            portfolio_weightage = np.random.randint(20, 40)
            avg_salary_lpa = round(np.random.uniform(2.0, 3.5), 2)

        else:
            # Core Tech Roles evolving with AI tools
            vacancies = np.random.randint(100, 180)
            hiring_difficulty = np.random.choice(['Medium', 'Hard'], p=[0.3, 0.7])
            degree_importance = np.random.choice(['High', 'Medium', 'Low'], p=[0.2, 0.5, 0.3])
            portfolio_weightage = np.random.randint(65, 90)

            if role == 'Web Developer':
                top_skill = 'React JS & AI Copilot'
            elif role == 'Data Analyst':
                top_skill = 'Python, SQL & Power BI'
            else:
                top_skill = 'Cloud Computing & System Design'

            skill_demand_status = 'Growing'
            avg_salary_lpa = round(np.random.uniform(6.0, 12.0), 2)

    # Append structured row data
    dataset.append([
        year, role, vacancies, hiring_difficulty,
        top_skill, skill_demand_status, degree_importance,
        portfolio_weightage, avg_salary_lpa
    ])

# Create Structured DataFrame
columns = [
    'Year', 'Job_Role', 'Vacancies', 'Hiring_Difficulty',
    'Top_Required_Skill', 'Skill_Demand_Status', 'Degree_Importance',
    'Portfolio_Weightage_Percentage', 'Avg_Salary_LPA'
]
df = pd.DataFrame(dataset, columns=columns)

# Export to CSV
df.to_csv('ai_job_impact_data.csv', index=False)
print("Dataset build complete! Saved successfully as 'ai_job_impact_data.csv'.")
```
